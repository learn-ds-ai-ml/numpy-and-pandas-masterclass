# random number generation

The modern way to generate random numbers in NumPy is to use **`numpy.random.default_rng()`** to create a Generator object. This updated API, introduced in NumPy 1.17, is faster and more statistically secure than the legacy **`numpy.random`** functions.

Here is a quick-start guide to generating different types of random data.

## 🚀 Quick Examples

```python
import numpy as np

# intialize the generator(do this once)
rng = np.random.default_rng()

# 1. Random flaots bw 0.0 and 1.0
floats = rng.random(size=5)

# 2. Random integers bw a range (eg., 1 to 10)
integers = rng.integers(low=1, high=11, size=5)

# 3. Numbers from a Normal (Gaussian) distribution
normal_dist = rng.normal(loc=0.0, scale=1.0, size=5)

print(f"random floats: {floats}")
print(f"random itegers: {integers}")
print(f"random numbers from normal distribution: {normal_dist}")
```

## 📊 Common Generator Methods
The Generator object provides specific methods depending on the type of data you need:

- **`rng.random(size)`:** Generates continuous floats in the semi-open interval `[0.0, 1.0)`.
- **`rng.integers(low, high, size, endpoint=False)`:** Generates random integers. Set `endpoint=True` to include the `high` value.
- **`rng.choice(a, size, replace=True, p=None)`:** Generates a random sample from a 1D array or list `a`. You can optionally provide probabilities `p` for each element.
- **`rng.shuffle(x)`:** Modifies a sequence in-place by shuffling its contents.
- **`rng.permutation(x)`:** Returns a randomly permuted sequence or array without altering the original data.


------------------------------

## 🧬 Statistical Distributions

NumPy supports dozens of probability distributions. Here are three of the most widely used:

| **Distribution** | **Method** | **Key Parameters** |
|---|---|---|
| **Normal / Gaussian** | `rng.normal()` | `loc` (mean), `scale` (standard deviation) |
| **Uniform** | `rng.uniform()` | `low` (minimum), `high` (maximum) |
| **Binomial** | `rng.binomial()` | `n` (trials), `p` (probability of success) |


Example of generating a 3 × 3 matrix from a Uniform distribution:


```python
rng.uniform(low=10.0, high=20.0, size=(3,3))
```

------------------------------
## 🔒 Reproducibility (Seeding)

To make your random numbers predictable across script runs (critical for testing and debugging), pass an integer **seed** to the constructor:


```python
# Always produces the exact same of sequence of numbers
rng2 = np.random.default_rng(seed=42)
rng2.random()
```

------------------------------
## ⚠️ Legacy vs. Modern API

You will often see older tutorials use `np.random.seed()` and functions like `np.random.rand()` or `np.random.randint()`.

- **Legacy:** `np.random.randint(1, 10, 5)`
- **Modern:** `rng.integers(1, 10, 5)`

**Why use the modern API?** It uses a superior pseudo-random number generator algorithm (PCG64 instead of MT19937), offers better performance, and is completely thread-safe.
